Task 3 – Randomness and Ludo Dice Simulation

```python
print("Task 03:")
# Generate one random value using seed 2026
np.random.seed(2026)

random_value = np.random.rand()
print(" Random Value:", random_value)

# Generate 25 normally distributed random values
random_numbers = np.random.randn(25)
print("25 Random Numbers:\n", random_numbers)

# Set seed for reproducibility
np.random.seed(42)

# Simulate 10,000 rolls of two dice
dice = np.random.randint(1, 7, size=(10000, 2))

# Count occurrences of each dice face
faces, counts = np.unique(dice, return_counts=True)

print(" Dice Faces:", faces)
print(" Face Counts:", counts)

# Calculate probability of getting a six
probability_six = np.mean(dice == 6)

print(" Probability of Six:", probability_six)

# Count double sixes
double_sixes = np.sum(np.all(dice == 6, axis=1))

print(" Number of Double Sixes:", double_sixes)
```

Task 6 – Pakistan Census 2023 Analysis

```python
print("Task 06:")
# Population units
units = np.array([
    "Punjab",
    "Sindh",
    "Khyber Pakhtunkhwa",
    "Balochistan",
    "Islamabad",
    "AJK",
    "Gilgit-Baltistan"
])

# Population data for 2023
pop_2023 = np.array([
    127688922, 55696147, 40856097,
    14894402, 2363863, 4333467, 1709049
])

# Population data for 2017
pop_2017 = np.array([
    109989655, 47854510, 35501964,
    12335129, 2003368, 4045367, 1492924
])

# Calculate total population
total_population = pop_2023.sum()

print("Total Population:", total_population)

# Calculate percentage share of each unit
population_share = (pop_2023 / total_population) * 100

print("Population Share (%):")
print(np.round(population_share, 2))

# Find the most populous unit
most_populous = units[pop_2023.argmax()]

print("Most Populous Unit:", most_populous)

# Find units with population above 20 million
large_units = units[pop_2023 > 20_000_000]

print(" Units Above 20 Million:", large_units)

# Calculate annual population growth from 2017 to 2023
annual_growth = ((pop_2023 / pop_2017) ** (1 / 6) - 1) * 100

print(" Annual Growth Rate (%):")
print(np.round(annual_growth, 2))

# Find the fastest-growing unit
fastest_growth = units[annual_growth.argmax()]

print("Fastest Growing Unit:", fastest_growth)
```

Task 7 – Climate Data Analysis

```python
print("Task 07:")
# City and month names
cities = np.array([
    "Lahore",
    "Karachi",
    "Islamabad",
    "Peshawar",
    "Quetta"
])

months = np.array([
    "Jan", "Feb", "Mar", "Apr",
    "May", "Jun", "Jul", "Aug",
    "Sep", "Oct", "Nov", "Dec"
])

# Maximum temperature data in Celsius
tmax = np.array([
    [18.4, 22.2, 27.5, 34.2, 38.9, 38.9, 35.6, 34.7, 34.4, 32.4, 27.1, 21.4],
    [26.3, 28.7, 32.6, 35.0, 35.7, 35.7, 33.6, 32.5, 33.4, 35.6, 32.6, 28.4],
    [17.7, 20.0, 24.8, 30.6, 36.1, 38.3, 35.4, 33.9, 33.4, 30.9, 25.4, 20.4],
    [18.1, 20.1, 24.7, 30.6, 36.5, 39.6, 37.4, 35.9, 34.8, 31.0, 25.2, 20.6],
    [11.8, 14.0, 19.5, 25.9, 31.4, 35.5, 36.8, 35.4, 31.9, 26.0, 19.6, 14.6]
])

# Monthly rainfall data in millimetres
rain = np.array([
    [21.9, 31.5, 31.5, 19.5, 17.7, 73.8, 188.6, 177.1, 88.6, 10.3, 6.9, 6.8],
    [10.6, 5.5, 3.2, 11.1, 9.2, 24.6, 86.1, 104.7, 44.0, 12.7, 0.7, 5.6],
    [55.2, 93.4, 95.2, 58.1, 39.9, 78.4, 310.6, 317.0, 135.4, 34.4, 17.7, 25.9],
    [39.8, 69.9, 78.2, 68.1, 26.7, 25.4, 69.0, 68.5, 34.8, 26.7, 16.9, 17.3],
    [55.0, 58.4, 49.5, 40.0, 17.8, 3.2, 8.8, 4.9, 4.2, 4.5, 19.3, 34.6]
])

# Display the shape of temperature data
print(" Temperature Shape:", tmax.shape)

# Calculate yearly mean temperature for each city
yearly_mean = tmax.mean(axis=1)

print(" Yearly Mean Temperature (°C):")
print(np.round(yearly_mean, 2))

# Find the hottest month for each city
hottest_month_index = tmax.argmax(axis=1)

hottest_month = months[hottest_month_index]

print(" Hottest Month of Each City:")
print(hottest_month)

# Calculate average temperature across five cities for each month
monthly_average = tmax.mean(axis=0)

print("Monthly Average Temperature (°C):")
print(np.round(monthly_average, 2))

# Find temperatures above 35 Celsius
hot_mask = tmax > 35

print(" Temperatures Above 35°C:")
print(tmax[hot_mask])

# Display city and month positions above 35 Celsius
city_index, month_index = np.where(hot_mask)

print("Cities with Temperatures Above 35°C:")
print(cities[city_index])
print(" Corresponding Months:")
print(months[month_index])

# Convert Lahore temperature from Celsius to Fahrenheit
lahore_fahrenheit = tmax[0] * 9/5 + 32

print(" Lahore Temperature in Fahrenheit:")
print(np.round(lahore_fahrenheit, 2))

# Calculate monsoon rainfall (July, August, September)
monsoon_rain = rain[:, 6:9].sum(axis=1)

print(" Monsoon Rainfall (mm):")
print(monsoon_rain)

# Calculate annual rainfall
annual_rain = rain.sum(axis=1)

print(" Annual Rainfall (mm):")
print(annual_rain)

# Calculate monsoon rainfall percentage
monsoon_percentage = (monsoon_rain / annual_rain) * 100

print(" Monsoon Rainfall Share (%):")
print(np.round(monsoon_percentage, 2))

# Find the city with the highest monsoon rainfall
wettest_city = cities[monsoon_rain.argmax()]

print(" Wettest Monsoon City:", wettest_city)
```

Task 8 – Inflation Data Analysis

```python
print("Task 08:")
# Define month names
inflation_months = np.array([
    "Jul", "Aug", "Sep", "Oct", "Nov",
    "Dec", "Jan", "Feb", "Mar", "Apr"
])

# Inflation data from the lab
cpi = np.array([
    [28.3, 27.4, 31.4, 26.8, 29.2,
     29.7, 28.3, 23.1, 20.7, 17.3],

    [11.1, 9.6, 6.9, 7.2, 4.9,
     4.1, 2.4, 1.5, 0.7, 0.3]
])

# Calculate average inflation for each financial year
average_inflation = cpi.mean(axis=1)

print(" Average Inflation (%):")
print(np.round(average_inflation, 2))

# Find FY2024 months with inflation above 25 percent
high_inflation = inflation_months[cpi[0] > 25]

print(" FY2024 Months Above 25%:")
print(high_inflation)

# Calculate monthly inflation drop
inflation_drop = cpi[0] - cpi[1]

print(" Monthly Inflation Drop (%):")
print(np.round(inflation_drop, 2))

# Find the month with the biggest inflation drop
biggest_drop_index = inflation_drop.argmax()

print("T8 - Month of Biggest Drop:",
      inflation_months[biggest_drop_index])

print(" Biggest Drop (%):",
      inflation_drop[biggest_drop_index])

# Classify FY2025 inflation using np.where
# Low: below 5%, Moderate: 5% to below 10%, High: 10% or above

inflation_category = np.where(
    cpi[1] < 5,
    "Low",
    np.where(cpi[1] < 10, "Moderate", "High")
)

print(" FY2025 Inflation Categories:")
print(inflation_category)
```

Task 9 – Grace Marks Analysis

```python
print("Task 09:")


# Original marks of students
marks = np.array([45, 67, 32, 89, 50, 38, 76, 95, 42, 60])

# Add 5 grace marks and cap maximum marks at 100
adjusted_marks = np.clip(marks + 5, 0, 100)

print("Original Marks:", marks)
print("Adjusted Marks:", adjusted_marks)

# Identify students who passed (50 or above)
passed = adjusted_marks >= 50
print("Passed Students:", passed)
print("Number of Passed Students:", np.sum(passed))

# Identify students who passed only because of grace marks
grace_pass = (marks < 50) & (adjusted_marks >= 50)
print("Students Passed Due to Grace Marks:", grace_pass)
print("Number of Grace Pass Students:", np.sum(grace_pass))

# Calculate mean and standard deviation
mean_marks = np.mean(adjusted_marks)
std_marks = np.std(adjusted_marks)

print("Mean Marks:", mean_marks)
print("Standard Deviation:", std_marks)

# Calculate Z-scores
z_scores = (adjusted_marks - mean_marks) / std_marks
print("Z-Scores:", z_scores)
```

Task 10 – Image is an Array

```python
# Create an 8x8 image filled with zeros
img = np.zeros((8, 8), dtype=int)

# Draw the number 7
img[1, 1:7] = 255
img[2:7, 5] = 255
img[4, 3:7] = 128

print(" Original Image:\n", img)

# Flip image horizontally
flipped_image = img[:, ::-1]

print(" Horizontally Flipped Image:\n", flipped_image)

# Flip image vertically
vertical_image = img[::-1, :]

print(" Vertically Flipped Image:\n", vertical_image)

# Invert image colours
inverted_image = 255 - img

print(" Inverted Image:\n", inverted_image)

# Crop the image
cropped_image = img[1:5, 3:7]

print(" Cropped Image:\n", cropped_image)

# Increase brightness and cap values at 255
bright_image = np.clip(img + 80, 0, 255)

print(" Brightened Image:\n", bright_image)

# Display image properties
print(" Image Shape:", img.shape)

print(" Image Data Type:", img.dtype)

# Count white pixels
white_pixels = np.sum(img == 255)

print(" Number of White Pixels:", white_pixels)
```
