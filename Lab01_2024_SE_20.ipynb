Task 10

```python
def caught_speeding(speed, is_birthday):
    if is_birthday:
        speed -= 5

    if speed <= 60:
        return "No Ticket"
    elif speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"

print("Output of task 10 is:\n", caught_speeding(60, False))
print(caught_speeding(75, False))
print( caught_speeding(90, False))
print( caught_speeding(65, True))
```
